```python
import math

N = 10
p = 0.5

def binom_pmf(n, k, p):
    return math.comb(n, k) * (p**k) * ((1-p)**(n-k))

# normalization constant
Z = sum(binom_pmf(N, k, p) for k in range(6, N+1))

# conditional distribution
pmf_cond = {l: binom_pmf(N, l, p)/Z for l in range(6, N+1)}

for l, prob in pmf_cond.items():
    print(l, prob)
```
